```python
import numpy as np

# Mini Project: Student Score Analysis with NumPy

np.random.seed(42)

# 1) Create synthetic data: 20 students, 5 subjects
scores = np.random.randint(40, 101, size=(20, 5))

# 2) Add some missing values manually
scores = scores.astype(float)
scores[2, 1] = np.nan
scores[7, 3] = np.nan
scores[15, 0] = np.nan

print("Original Scores:\n", scores)

# 3) Detect missing values
missing_mask = np.isnan(scores)
print("\nMissing Value Mask:\n", missing_mask)

# 4) Replace missing values with subject-wise mean
subject_means = np.nanmean(scores, axis=0)
rows, cols = np.where(np.isnan(scores))
scores[rows, cols] = subject_means[cols]

print("\nCleaned Scores:\n", scores)
print("\nSubject Means:", subject_means)

# 5) Basic statistics
student_totals = np.sum(scores, axis=1)
student_averages = np.mean(scores, axis=1)
subject_max = np.max(scores, axis=0)
subject_min = np.min(scores, axis=0)
subject_std = np.std(scores, axis=0)

print("\nStudent Totals:\n", student_totals)
print("\nStudent Averages:\n", np.round(student_averages, 2))
print("\nSubject Max:", subject_max)
print("Subject Min:", subject_min)
print("Subject Std:", np.round(subject_std, 2))

# 6) Rank students by average
rank_indices = np.argsort(student_averages)[::-1]
ranked_averages = student_averages[rank_indices]

print("\nStudent Ranking (best to worst):")
for i, idx in enumerate(rank_indices):
    print(f"Rank {i+1}: Student {idx+1} -> Avg = {student_averages[idx]:.2f}")

# 7) Find top performers
top_students = np.where(student_averages >= 85)[0] + 1
print("\nTop Students (average >= 85):", top_students)

# 8) Pass/fail analysis
pass_matrix = np.where(scores >= 50, 1, 0)
students_passed_all = np.all(scores >= 50, axis=1)
students_failed_any = np.any(scores < 50, axis=1)

print("\nPass Matrix:\n", pass_matrix)
print("\nStudents who passed all subjects:", np.where(students_passed_all)[0] + 1)
print("Students who failed at least one subject:", np.where(students_failed_any)[0] + 1)

# 9) Normalize scores subject-wise using min-max scaling
subject_min_vals = np.min(scores, axis=0)
subject_max_vals = np.max(scores, axis=0)
normalized_scores = (scores - subject_min_vals) / (subject_max_vals - subject_min_vals)

print("\nNormalized Scores:\n", np.round(normalized_scores, 2))

# 10) Bonus insights
best_student_index = np.argmax(student_averages)
worst_student_index = np.argmin(student_averages)

print(f"\nBest Student: Student {best_student_index + 1} with average {student_averages[best_student_index]:.2f}")
print(f"Worst Student: Student {worst_student_index + 1} with average {student_averages[worst_student_index]:.2f}")

# 11) Subject correlation-like comparison using transpose and dot
# This is not true correlation, just a simple matrix interaction example
subject_interaction = np.dot(scores.T, scores)

print("\nSubject Interaction Matrix:\n", np.round(subject_interaction, 2))
```

```python
import numpy as np

# UniBrain Mini-Project: Medical Image Processing & Tumor Detection Simulator
# Goals: Noise reduction, Normalization, Segmentation, and Feature Extraction

# 1. شبیه‌سازی یک تصویر MRI (سایز 64x64 پیکسل) با مقادیر شدت روشنایی بین 0 تا 255
np.random.seed(7)
image = np.random.randint(0, 50, size=(64, 64), dtype=np.uint8) # نویز زمینه

# 2. ایجاد یک "تومور" فرضی (یک ناحیه روشن‌تر در مرکز تصویر)
# از np.indices برای پیدا کردن مختصات پیکسل‌ها استفاده می‌کنیم
y, x = np.indices((64, 64))
center_y, center_x = 30, 32
mask_area = np.sqrt((x - center_x)**2 + (y - center_y)**2) < 8
image[mask_area] = np.random.randint(180, 255, size=np.sum(mask_area))

print(f"--- Step 1: Image Created ---")
print(f"Image Shape: {image.shape} | Max Intensity: {np.max(image)}")

# 3. نرمال‌سازی (Normalization): تبدیل مقادیر به بازه [0, 1] برای ورود به مدل‌های Deep Learning
# فرمول: (x - min) / (max - min)
img_min, img_max = np.min(image), np.max(image)
normalized_img = (image - img_min) / (img_max - img_min)

# 4. حذف نویز (Basic Filtering): استفاده از میانگین متحرک ساده با استفاده از Clipping
# مقادیر بسیار پایین (نویز) را به صفر تبدیل می‌کنیم
denoised_img = np.where(normalized_img < 0.2, 0, normalized_img)

# 5. قطعه‌بندی (Segmentation): شناسایی ناحیه تومور با Thresholding
# هر پیکسلی که از 70% شدت عبور کند به عنوان بافت مشکوک در نظر گرفته می‌شود
threshold = 0.7
tumor_mask = denoised_img > threshold

# 6. استخراج ویژگی (Feature Extraction):
# الف) محاسبه مساحت تومور (تعداد پیکسل‌های مثبت)
tumor_area = np.sum(tumor_mask)

# ب) پیدا کردن مرکز ثقل تومور (Center of Mass)
tumor_coords = np.argwhere(tumor_mask) # پیدا کردن مختصات تمام پیکسل‌های تومور
center_of_mass = tumor_coords.mean(axis=0)

# ج) محاسبه میانگین شدت روشنایی در ناحیه تومور
avg_intensity = np.mean(image[tumor_mask])

print(f"\n--- Analysis Results ---")
print(f"Tumor Detected: {np.any(tumor_mask)}")
print(f"Tumor Area: {tumor_area} pixels")
print(f"Center of Mass (Y, X): {np.round(center_of_mass, 2)}")
print(f"Average Tumor Intensity: {avg_intensity:.2f}")

# 7. داده‌افزایی (Data Augmentation): تولید نسخه‌های جدید برای آموزش مدل
flipped_lr = np.fliplr(image) # چرخش افقی
rotated_90 = np.rot90(image, k=1) # چرخش 90 درجه
bright_img = np.clip(image.astype(float) * 1.2, 0, 255).astype(np.uint8) # افزایش روشنایی

# 8. دسته‌بندی نهایی (Stacking): ترکیب تصاویر برای نمایش در یک پکیج
augmented_batch = np.stack([image, flipped_lr, rotated_90, bright_img], axis=0)

print(f"\n--- Augmentation Info ---")
print(f"Batch Shape for CNN: {augmented_batch.shape} (Batch_size, H, W)")
print(f"Unique Intensities in Mask: {np.unique(image[tumor_mask])[:5]}... (Truncated)")

# 9. بررسی نهایی سلامت داده: چک کردن برای مقادیر Infinity یا NaN
is_clean = not np.any(np.isnan(normalized_img)) and not np.any(np.isinf(normalized_img))
print(f"Data Cleanliness Check: {'Passed' if is_clean else 'Failed'}")
```
