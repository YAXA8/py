```python

arr = np.array([1.5, 2.7, 3.2, 4.9, 5.1])
filtered = np.where(arr > 3, 1, 0)
print(filtered)
```

```python
unique_vals = np.unique(arr)
unique_vals
```

```python
expanded = np.expand_dims(arr, axis=0)
print(expanded)
```
